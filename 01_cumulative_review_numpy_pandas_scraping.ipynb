```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("./data/travel.csv")
df.dtypes
df.isnull().sum()
df.duplicated().sum()
df[df.duplicated(keep=False)]
df.dropna(subset=["상세영업상태","도로명"], inplace=True)
df.drop_duplicates(keep="first", inplace=True)
df.isnull().sum()
df[df.duplicated(keep=False)]
df.drop('도로명', axis=1, inplace=True)
df.dtypes
```

```python
df['자본금'].describe()
df = df[df['상세영업상태'] =="영업중"]
df.loc[df['자본금'] < 30000000, '자본금'] = 32000000
df['인허가일자'] = pd.to_datetime(df['인허가일자'])
df['개점년도'] = df['인허가일자'].dt.year
```

```python
cnt = df.groupby('개점년도')['사업장명'].count()
num = df.groupby('개점년도')['자본금'].sum()
grp = pd.merge(cnt,num, on="개점년도", how ="inner")
grp.rename(columns = {'사업장명' : "년도별사업장개수"}, inplace=True)
grp['순위'] = grp['년도별사업장개수'].rank(method="dense", ascending = False)
grp['순위']= grp['순위'].astype(int)
grp.sort_values(by='순위', inplace=True, ascending = True)
brank_rank = grp[(grp.index >2018) & (grp.index <= 2023)]
brank_rank.sort_index(ascending =True, inplace=True)
brank_rank
```

```python
plt.rcParams['font.family'] = 'malgun gothic'
plt.figure(figsize=(8,6), facecolor = "w")
plt.title("전국 여행사 대리점 년도별 개점 현황", color ="g", fontsize=20)
sns.set_theme(style="white")
color = ['b','orange','g','white','yellow']
plt.bar(brank_rank.index, brank_rank['년도별사업장개수'], color=color, linewidth = 4, edgecolor = "r", hatch = "\\\\", alpha =0.5)
plt.xlabel("year", color ="b", fontsize=8)
plt.ylabel("매장개수", color ="b", fontsize=8)
plt.draw()
plt.savefig("./images/travel.jpeg", dpi=300)
```
